# CONCLUSION

This project trained a Logistic Regression model on NASA close approach data to classify asteroids as potentially hazardous (impact) or safe (halo), using **NASA's official PHO criteria** as the target definition.

The final model correctly identified 100% of potentially hazardous asteroids in the test set with an overall accuracy of 95%. Random Forest was also tested but discarded because it predicted the majority class for every row, making its high accuracy misleading and useless for hazard detection.

**One honest limitation:** the target column was engineered using the same features used for training (distance and magnitude). This means the model may not generalize well to asteroid data from different sources or with different feature distributions.
Given more time, the next steps would be testing additional models like Decision Trees or Support Vector Machines, addressing class imbalance using SMOTE, and building an interactive Streamlit app to make the model accessible to a general audience.
